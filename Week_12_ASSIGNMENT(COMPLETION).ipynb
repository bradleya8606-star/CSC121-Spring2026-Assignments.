```python
# NAME: rectangle.py
# IMPORT STATEMENTS
# (None needed for basic calculations)

# FUNCTION DEFINITIONS
def calculate_rectangle(length, width):
    """
    Calculate the area and perimeter of a rectangle.

    Parameters:
    length (int or float): Length of the rectangle
    width (int or float): Width of the rectangle

    Returns:
    tuple: (area, perimeter)
    """
    area = length * width
    perimeter = 2 * (length + width)
    return (area, perimeter)

# PROGRAM LOGIC
# Get user input
length = float(input("Enter the length of the rectangle: "))
width = float(input("Enter the width of the rectangle: "))

# Call the function and unpack the tuple
area, perimeter = calculate_rectangle(length, width)

# Print results clearly
print(f"Area of the rectangle: {area}")
print(f"Perimeter of the rectangle: {perimeter}")
```
