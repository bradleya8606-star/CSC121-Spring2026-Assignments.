```python
# IMPORT STATEMENTS
# (None)

# FUNCTION DEFINITIONS
def calculate_ticket(age, hour):
    # Base price
    if age <= 12 or age >= 65:
        price = 8
    else:
        price = 12

    # Matinee discount
    if hour < 17:
        price -= 2

    # Print final price
    print("Final ticket price: $", price)


# PROGRAM LOGIC
age = int(input("Enter age: "))
hour = int(input("Enter movie hour (0-23): "))

calculate_ticket(age, hour)
```
