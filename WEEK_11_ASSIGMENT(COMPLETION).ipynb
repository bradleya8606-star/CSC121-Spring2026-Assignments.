```python
# NAME: gradebook.py
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS
def add_students(gradebook, name, grade):
    gradebook[name] = [grade]


def calculate_average(grades):
    if len(grades) == 0:
        return 0
    return sum(grades) / len(grades)


# PROGRAM LOGIC
def main():
    gradebook = {}

    # Input loop
    while True:
        name = input("Enter student name (or 'done'): ").strip()

        if name.lower() == "done":
            break

        grade = float(input(f"Enter {name}'s grade: "))
        add_students(gradebook, name, grade)

    # Final reporting logic
    if len(gradebook) == 0:
        print("No students were entered.")
    else:
        print("\nGrade Report:")

        total_sum = 0
        total_count = 0

        for student, grades in gradebook.items():
            avg = calculate_average(grades)
            print(f"{student}: Average = {avg:.2f}")

            # Add to class totals
            total_sum += sum(grades)
            total_count += len(grades)

        # Calculate class average safely
        if total_count == 0:
            class_avg = 0
        else:
            class_avg = total_sum / total_count

        print(f"\nClass Average: {class_avg:.2f}")


if __name__ == "__main__":
    main()
```
