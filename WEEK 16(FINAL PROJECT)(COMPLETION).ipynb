```python
# NAME: expense_tracker.py

expenses = []

while True:
    print("\n--- Expense Tracker ---")
    print("1. Add New Expense")
    print("2. Get Totals")
    print("3. Exit")

    choice = input("Selection: ")

    if choice == "1":
        category = input("Enter Category: ")
        vendor = input("Enter Vendor: ")
        date = input("Enter Date: ")

        while True:
            try:
                amount = float(input("Enter Amount: "))
                break
            except ValueError:
                print("Invalid amount. Please enter a number.")

        expense = {
            "category": category,
            "vendor": vendor,
            "amount": amount,
            "date": date
        }

        expenses.append(expense)
        print("Expense added!")

    elif choice == "2":
        print("\n--- Totaling Options ---")
        print("A: All Expenses")
        print("C: Filter by Category")
        print("V: Filter by Vendor")

        option = input("Choice: ").upper()

        total = 0

        if option == "A":
            for expense in expenses:
                total += expense["amount"]
            print(f"Total of all expenses: ${total:.2f}")

        elif option == "C":
            cat = input("Enter Category to sum: ")
            for expense in expenses:
                if expense["category"].lower() == cat.lower():
                    total += expense["amount"]
            print(f"The total for {cat} is: ${total:.2f}")

        elif option == "V":
            ven = input("Enter Vendor to sum: ")
            for expense in expenses:
                if expense["vendor"].lower() == ven.lower():
                    total += expense["amount"]
            print(f"The total for {ven} is: ${total:.2f}")

    elif choice == "3":
        print("Goodbye!")
        break
```
