```python
# FUNCTION DEFINITIONS

def add_players(roster):
    while True:
        name = input("Enter player name (or 'quit' to stop): ")

        if name.lower() == "quit":
            break

        roster.append(name)


def is_on_roster(roster, search_name):
    for player in roster:
        if player.lower() == search_name.lower():
            return True

    return False


# PROGRAM LOGIC

roster = []

add_players(roster)

search_name = input("Who would you like to search for? ")

found = is_on_roster(roster, search_name)

if found:
    print("Result: Player found!")
else:
    print("Result: Player not found.")
```
