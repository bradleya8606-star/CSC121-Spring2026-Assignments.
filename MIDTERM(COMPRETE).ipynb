```python
# NAME: rentalcar.py

# ---------------------------------------
# IMPORT STATEMENTS
# ---------------------------------------
# (No imports required for this program)


# ---------------------------------------
# FUNCTION DEFINITIONS
# ---------------------------------------

# Constants
DAILY_RATES = {
    "B": 40,
    "S": 60,
    "L": 80
}

CLASSIFICATION_NAMES = {
    "B": "Budget",
    "S": "Standard",
    "L": "Luxury"
}

FREE_MILES = 100
MILEAGE_RATE = 0.35


def validate_classification(code):
    """
    Validates the classification code.
    Returns the uppercase code if valid, otherwise None.
    """
    code = code.upper()
    if code in DAILY_RATES:
        return code
    return None


def calculate_mileage_charge(total_miles):
    """
    Calculates how many miles are charged and the mileage cost.
    Returns (charged_miles, mileage_cost).
    """
    charged_miles = max(0, total_miles - FREE_MILES)
    mileage_cost = charged_miles * MILEAGE_RATE
    return charged_miles, mileage_cost


def calculate_total_bill(classification_code, rental_days, total_miles):
    """
    Calculates the full rental bill.
    NOTE: No print statements inside this function.
    Returns a dictionary with all billing details.
    """
    daily_rate = DAILY_RATES[classification_code]
    base_cost = daily_rate * rental_days

    charged_miles, mileage_cost = calculate_mileage_charge(total_miles)

    total_bill = base_cost + mileage_cost

    return {
        "classification_name": CLASSIFICATION_NAMES[classification_code],
        "total_miles": total_miles,
        "charged_miles": charged_miles,
        "total_bill": total_bill
    }


# ---------------------------------------
# PROGRAM LOGIC
# ---------------------------------------

def main():
    # Validate classification input
    while True:
        code_input = input("Enter car classification (B, S, L): ")
        valid_code = validate_classification(code_input)

        if valid_code:
            break
        else:
            print("Invalid classification code. Please enter B, S, or L.")

    # Get rental details
    rental_days = int(input("Enter number of rental days: "))
    total_miles = float(input("Enter total miles driven: "))

    # Calculate bill
    bill = calculate_total_bill(valid_code, rental_days, total_miles)

    # Display final output
    print("\n------ Rental Summary ------")
    print("Car Classification:", bill["classification_name"])
    print("Total Miles Driven:", bill["total_miles"])
    print("Miles Charged:", bill["charged_miles"])
    print("Total Bill: ${:.2f}".format(bill["total_bill"]))


if __name__ == "__main__":
    main()
```
