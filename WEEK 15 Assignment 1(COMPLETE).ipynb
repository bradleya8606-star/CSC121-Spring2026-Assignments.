```python
# NAME: task_manager.py

from dataclasses import dataclass, field
from datetime import datetime

HISTORY_FILE = "completed_tasks.txt"


@dataclass
class Task:
    name: str
    priority: int  # 1 = High, 2 = Medium, 3 = Low
    created_at: datetime = field(default_factory=datetime.now)

    def __str__(self):
        priority_map = {1: "High", 2: "Medium", 3: "Low"}
        return f"[{priority_map[self.priority]}] {self.name}"


def add_task(task_list):
    name = input("Enter task name: ").strip()

    while True:
        try:
            priority = int(input("Enter priority (1=High, 2=Medium, 3=Low): "))
            if priority in (1, 2, 3):
                break
            else:
                print("Please enter 1, 2, or 3.")
        except ValueError:
            print("Invalid input. Enter a number.")

    task_list.append(Task(name, priority))
    print("Task added successfully!\n")


def view_tasks(task_list):
    if not task_list:
        print("No tasks added yet.\n")
        return

    print("\n--- Your Tasks ---")
    sorted_tasks = sorted(task_list, key=lambda t: (t.priority, t.created_at))
    for idx, task in enumerate(sorted_tasks, start=1):
        print(f"{idx}. {task}")
    print()


def complete_task(task_list):
    if not task_list:
        print("No tasks to complete.\n")
        return

    sorted_tasks = sorted(task_list, key=lambda t: (t.priority, t.created_at))

    print("\n--- Complete a Task ---")
    for idx, task in enumerate(sorted_tasks, start=1):
        print(f"{idx}. {task}")

    while True:
        try:
            choice = int(input("Select task number to complete: "))
            if 1 <= choice <= len(sorted_tasks):
                task_to_remove = sorted_tasks[choice - 1]
                break
            else:
                print("Invalid number.")
        except ValueError:
            print("Enter a valid number.")

    task_list.remove(task_to_remove)

    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    with open(HISTORY_FILE, "a") as file:
        file.write(f"{timestamp} - {task_to_remove.name}\n")

    print(f"Task '{task_to_remove.name}' completed and saved to history!\n")


def what_is_next(task_list):
    # Filter for Priority 1 tasks
    high_priority_tasks = [t for t in task_list if t.priority == 1]

    if not high_priority_tasks:
        print("\n🎉 No High priority tasks! You're doing great. Keep the momentum!\n")
        return

    # Find the oldest one
    oldest_task = min(high_priority_tasks, key=lambda t: t.created_at)

    print("\n🔥 What is next?")
    print(f"Focus on this task:\n{oldest_task}\n")


def main():
    tasks = []

    while True:
        print("=== Personal Task Manager ===")
        print("1. Add Task")
        print("2. View Tasks")
        print("3. Complete Task")
        print("4. What is next?")
        print("5. Exit")

        choice = input("Choose an option: ").strip()

        if choice == "1":
            add_task(tasks)
        elif choice == "2":
            view_tasks(tasks)
        elif choice == "3":
            complete_task(tasks)
        elif choice == "4":
            what_is_next(tasks)
        elif choice == "5":
            print("Goodbye!")
            break
        else:
            print("Invalid choice. Try again.\n")


if __name__ == "__main__":
    main()
```
